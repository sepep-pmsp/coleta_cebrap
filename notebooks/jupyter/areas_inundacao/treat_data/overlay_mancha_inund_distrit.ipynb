# Intersecção (overlay intersection) entre os Distritos e as Manchas de Inundação

```python
data_path= get_data_diretorio()
assets_path = os.path.join(
    data_path,
    'assets'
)
```

# Dependências

Este notebook é dependente dos parquets resultantes dos notebooks "../download_data/mancha_inundacao_25" e "../../arborizacao_viaria/malha_distritos"

```python
gdf_distrito = gpd.read_parquet(
    os.path.join(
        assets_path,
        'distrito_ibge.parquet'
    )
)
```

```python
gdf_mancha_inund = gpd.read_parquet(
    os.path.join(
        data_path,
        'assets',
        'mancha_inundacao_25.parquet'
    )
)
```

```python
gdf_mancha_inund.shape
```

# Overlay Intersection

```python
overlay_inund_dist = gpd.overlay(
    gdf_mancha_inund,
    gdf_distrito,
    how='intersection',
    keep_geom_type=True
)
```

```python
overlay_inund_dist.shape
```

```python
overlay_inund_dist.sample(2)
```

```python
overlay_inund_dist.columns
```

```python
overlay_inund_dist['area_inund_recort']= overlay_inund_dist['geometry'].area
```

# Salvar GDF

```python
save_parquet_excel(
    overlay_inund_dist,
    'overlay_mancha_inund_dist',
    assets_path,
    data_subpath = 'areas_inundacao'
)
```
